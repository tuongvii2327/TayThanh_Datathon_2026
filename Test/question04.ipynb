```python
import pandas as pd
```

```python
df_traffic = pd.read_csv('../dataset1/web_traffic.csv')
df_traffic.info()
print(df_traffic.head())
```

```python
df_traffic = df_traffic[['bounce_rate', 'traffic_source']]
df_traffic.info()
print(df_traffic['traffic_source'].value_counts())
```

```python
mean_by_trafficSource = df_traffic.groupby('traffic_source')['bounce_rate'].mean()
min_source = mean_by_trafficSource.idxmin()
min_mean = mean_by_trafficSource.min()

print(mean_by_trafficSource)
print(f"Nguồn truy cập: {min_source} có tỷ lệ thoát trung bình {min_mean} thấp nhất")
```
