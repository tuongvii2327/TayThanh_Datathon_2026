```python
import pandas as pd
```

```python
df_customer = pd.read_csv('../dataset1/customers.csv')
df_order = pd.read_csv('../dataset1/orders.csv')
df_customer.info()
df_order.info()
```

```python
df_customer = df_customer[['customer_id', 'age_group']]
df_order = df_order[['customer_id', 'order_id']]
print(df_customer['age_group'].value_counts())
```

```python
df_order = df_order.merge(df_customer, on='customer_id', how='left')
print(df_order)
```

```python
mean_by_age_group = df_order.groupby('age_group')['order_id'].count() / df_order.groupby('age_group')['customer_id'].nunique()
max_age_group = mean_by_age_group.idxmax()
max_mean = mean_by_age_group.max()
print(f"{max_age_group}: {max_mean}")
#print(mean_by_age_group)
```
