```python
import pandas as pd
```

```python
df_order = pd.read_csv('../dataset1/orders.csv')
df_pay = pd.read_csv('../dataset1/payments.csv')
#df_ret = pd.read_csv('../dataset1/returns.csv')
df_geo = pd.read_csv('../dataset1/geography.csv')
#df_item = pd.read_csv('../dataset1/order_items.csv')

df_order.info()
df_pay.info()
#df_ret.info()
df_geo.info()
#df_item.info()
```

```python
df_order = df_order[['order_id', 'order_date', 'zip', 'order_status']]
df_pay = df_pay[['order_id', 'payment_value']]
#df_ret = df_ret[['return_id', 'order_id', 'return_date', 'refund_amount']]
df_geo = df_geo[['zip', 'region']]

df_order['order_date'] = pd.to_datetime(df_order['order_date'], format='%Y-%m-%d')
#df_ret['return_date'] = pd.to_datetime(df_ret['return_date'], format='%Y-%m-%d')

start_time = pd.to_datetime('2012-07-04', format='%Y-%m-%d')
end_time = pd.to_datetime('2022-12-31', format='%Y-%m-%d')

df_order = df_order[(df_order['order_date'] >= start_time) & (df_order['order_date'] <= end_time)]
#df_ret = df_ret[(df_ret['return_date'] >= start_time) & (df_ret['return_date'] <= end_time)]

#print(df_order['order_status'].value_counts())

#df_order.info()
#df_ret.info()
#df_geo.info()
```

```python
#df_order = df_order[df_order['order_status'] != 'cancelled']

df_order = df_order.merge(df_geo, on='zip', how='left')
df_order = df_order.merge(df_pay, on='order_id', how='left')
#df_order = df_order.merge(df_ret, on='order_id', how='left')
#df_ret = df_ret.merge(df_order, on='order_id', how='left')

df_order.info()
```

```python

gross_revenue = df_order.groupby('region')['payment_value'].sum()
#return_order = df_ret.groupby('order_date')['refund_amount'].sum()
revenue = gross_revenue
#revenue = gross_revenue.sub(return_order, fill_value=0)
print(revenue)
```
