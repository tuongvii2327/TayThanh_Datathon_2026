```python
import pandas as pd
```

```python
df_return = pd.read_csv('../dataset1/returns.csv')
df_product = pd.read_csv('../dataset1/products.csv')
df_return.info()
df_product.info()
```

```python
print(df_product['category'].value_counts())
print(df_return['return_reason'].value_counts())
```

```python
df_product = df_product[['product_id', 'category']]
df_return = df_return[['return_id', 'product_id', 'return_reason']]
#cach 1, how='left' để đảm bảo không bị mất bất kỳ dòng dữ liệu trả hàng nào của bảng bên trái (df_return).
df_return = df_return.merge(df_product, on='product_id', how='left')
#cach 2
#mapping_dict = df_product.set_index('product_id')['category']
#df_return['category'] = df_return['product_id'].map(mapping_dict)
print(df_return)
print(df_product)
```

```python
df_return = df_return[df_return['category'] == 'Streetwear']
print(df_return)
```

```python
# Chú ý: mode() trả về một mảng (Series) vì có thể có 2 lý do cùng số lượng cao nhất.
# Thêm [0] để lấy phần tử đầu tiên (đáp án duy nhất).
mode_reason = df_return['return_reason'].mode()[0]

print("Lý do trả hàng nhiều nhất là:", mode_reason)
```
