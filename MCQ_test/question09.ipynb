```python
import pandas as pd
```

```python
df_ret = pd.read_csv('../dataset1/returns.csv')
df_item = pd.read_csv('../dataset1/order_items.csv')
df_product = pd.read_csv('../dataset1/products.csv')

df_ret.info()
df_item.info()
df_product.info()
```

```python
df_ret = df_ret[['return_id', 'order_id', 'product_id', 'return_quantity']]
df_item = df_item[['order_id' ,'product_id', 'quantity']]
df_product = df_product[['product_id', 'size']]

df_ret = df_ret.merge(df_product, on='product_id', how='left')
df_item = df_item.merge(df_product, on='product_id', how='left')

print(df_ret)
print(df_item)
```

```python
sale = df_item.groupby('size').size()
returns = df_ret.groupby('size').size()
print(f"{(returns / sale).idxmax()}: {(returns / sale).max()}")
print(returns / sale)
```
