```python
import pandas as pd
```

```python
df_pay = pd.read_csv('../dataset1/payments.csv')
df_pay.info()
```

```python
print(df_pay['installments'].value_counts())
```

```python
df_pay['installments'] = df_pay['installments'].astype('category')
sum_payment = df_pay.groupby('installments')['payment_value'].sum()
sum_order = df_pay.groupby('installments').size()
average = sum_payment / sum_order
print(f"{average.idxmax()}: {average.max()}")
print(average)
```
