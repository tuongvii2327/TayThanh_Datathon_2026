```python
import pandas as pd
```

```python
df_product = pd.read_csv('../dataset1/products.csv')
df_product.info()
```

```python
print(df_product['segment'].value_counts())
```

```python
df_product = df_product[['segment', 'price', 'cogs']]
df_product['segment'] = df_product['segment'].astype('category')
df_product['gross_profit_margin'] = (df_product['price'] - df_product['cogs']) / df_product['price']
print(df_product)
```

```python
# 1. Tính trung bình gross_profit_margin cho TẤT CẢ các segment cùng lúc
mean_by_segment = df_product.groupby('segment')['gross_profit_margin'].mean()

# 2. Lấy ra tên segment có giá trị lớn nhất (idxmax = Index of Maximum)
best_segment = mean_by_segment.idxmax()

# 3. (Tuỳ chọn) Lấy ra giá trị trung bình lớn nhất đó
best_mean = mean_by_segment.max()
print(mean_by_segment)
print(f"Phân khúc cao nhất: {best_segment} với tỷ suất {best_mean}")
```
